# 绘制2020年中国人民大学在云南省特定专业录取占比的环状饼图

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS', 'Heiti TC', 'Hiragino Sans GB', 'PingFang SC'] # 正常显示中文
plt.rcParams['axes.unicode_minus'] = False # 正常显示负号
```

```python
fig, axes = plt.subplots()
majors = ['法学', '工商管理', '国际政治', '金融学', '经济学', '人力资源管理', '新闻传播学']
numbers = [5, 10, 3, 8, 7, 3, 2]
colors = plt.color_sequences["Pastel2"]
axes.pie(x=numbers, labels=majors, colors=colors, autopct='%.1f %%', pctdistance =0.7)
axes.pie([1], colors=['white'], radius=0.5)
```
