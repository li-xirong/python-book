# 绘制2019年中国人民大学在北京市特定专业的录取分数柱状图（细化版）

设置y轴范围、显示数值

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS', 'Heiti TC', 'Hiragino Sans GB', 'PingFang SC'] # 正常显示中文
plt.rcParams['axes.unicode_minus'] = False # 正常显示负号
```

```python
fig, axes = plt.subplots()
majors = ['经济学', '新闻传播学', '人力资源管理', '国际政治', '外国语言文学']
scores = np.array([661, 664, 656, 654, 651]) 

bar = axes.bar(majors, scores, color='skyblue', width=0.6) 
axes.bar_label(bar, fontsize=14) # 在每根柱子上方添加对应的数值
axes.set_ylim(640, 670) # 设置y轴范围

axes.set_xlabel("专业", fontsize=14)
axes.set_ylabel("分数", fontsize=14)
```
