# 绘制2015-2023年中国人民大学部分专业在湖南省的文科录取分数线的热力图

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS', 'Heiti TC', 'Hiragino Sans GB', 'PingFang SC'] # 正常显示中文
plt.rcParams['axes.unicode_minus'] = False # 正常显示负号
```

```python
majors = ['工商管理', '法学', '金融学', '社会学', '国际政治', '经济学', '新闻传播学', '人力资源管理']
years = np.arange(2015, 2024)
scores = np.array([[630, 628, 655, 668, 651, 655, 640, 633, 648],
                   [634, 633, 658, 673, 656, 660, 649, 638, 654],
                   [635, 634, 654, 669, 653, 658, 645, 638, 649], 
                   [613, 611, 653, 647, 652, 653, 639, 638, 649], 
                   [611, 626, 654, 668, 650, 656, 652, 632, 648], 
                   [632, 631, 635, 668, 653, 656, 644, 636, 651], 
                   [628, 629, 656, 673, 653, 659, 648, 636, 654], 
                   [624, 626, 635, 667, 650, 654, 639, 632, 648]])
fig, axes = plt.subplots()
heatmap = axes.imshow(scores, cmap='coolwarm') 
cb = fig.colorbar(heatmap, ax=axes, aspect=20) 
axes.set_xticks(np.arange(years.shape[0])) 
axes.set_xticklabels(years) 
axes.set_yticks(np.arange(len(majors))) 
axes.set_yticklabels(majors)
```
