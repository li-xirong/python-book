## 示例代码 2.35 定义一个名为Person的类

````python
from datetime import date # 从 Python内置的datetime 模块中导入date类，用于计算时间

class Person:
    def __init__(self, birth_year, birth_month, birth_day):
        self.birth = (birth_year, birth_month, birth_day)

    def get_age(self):
        today = date.today()
        age = today.year - self.birth[0] 
        # 根据出生月份及日期对age进行必要的修正
        if (today.month < self.birth[1]) or (today.month == self.birth[1] and today.day < self.birth[2]):
            age = age - 1
        return age
        
    @staticmethod
    def is_adult(age):
        return age >= 18

p = Person(2001, 10, 1) # 创建一个该类的对象
print(f'年龄：{p.get_age()}') # 输出：年龄：24
age = p.get_age()
print(p.is_adult(age), Person.is_adult(age)) # 输出: True, True
````

## 示例代码2.36 定义一个Dog类并创建多个实例

````python
class Dog:
    def __init__(self, name, birth_year):
        self.name = name  
        self.birth_year = birth_year
    
    def bark(self):      # 类方法，无返回值
        print(f"{self.name} says: Woof!")

    def age(self): # 返回年龄
        age = date.today().year - self.birth_year
        return age

    def human_age(self): # 换算成人的年龄返回
        return self.age() * 7
# 创建类的实例（对象）
d1 = Dog('Fluppy', 2020)
d2 = Dog('Max', 2021)
# 访问属性和调用方法
print(d1.name, d2.name) # 输出 Fluppy, Max
d1.bark() # 输出 Fluppy says: Woof!
print(f'age {d2.age()}, human age {d2.human_age()}') # 输出: age 4, human age 28
````

## 示例代码2.37 简易员工信息管理程序

````python
class Employee(Person): # 继承
    def __init__(self, name, birth_date, role='普通员工'):
        year, month, day = list(map(int, birth_date.split('-')))
        super().__init__(year, month, day)
        self.name = name
        self.role = role

    def get_info(self):
        return f"姓名: {self.name}, 年龄: {self.get_age()}, 岗位: {self.role}"

class Manager(Employee): # 继承
    def __init__(self, name, birth_date, department):
        super().__init__(name, birth_date, '经理')
        self.department = department

    # 方法重写 (Override)
    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, 管理部门: {self.department}"

# 创建对象列表
staff = [
    Employee("小明", "2000-10-1"),
    Employee("小满", "1998-3-15"),
    Manager("老王", "1985-8-1", "技术部")
]
# 循环展示多态性
for person in staff:
    print(person.get_info())
````

## 示例代码2.44 在遍历列表时修改列表

````python
nums = [1, 2, 2, 3, 4]
for i,n in enumerate(nums):
    print(i, n)
    if n == 2:
        nums.remove(n)
print(nums)  # 期望 [1, 3, 4]，结果 [1, 2, 3, 4]
# 正确做法：列表推导式
nums = [n for n in nums if n!=2]
print(nums)
````

## 示例代码2.45 浅复制与深复制

````python
a = [[1,0], [0,1]]
b = list(a) # b是a的浅复制
b[0][0] = 2
print(a) # 输出：[[2, 0], [0, 1]]。a被修改了！
# 正确做法
import copy
b = copy.deepcopy(a) # 对b的任何修改都不会影响a
````

## 示例代码2.46 变量作用领

````python
count = 0 # 全局变量

def func1():
    print(count) # 不修改全局变量可以访问全局变量

def func2():
    count = count + 1 # 修改同名全局变量，则Python会认为它是一个局部变量，报错：UnboundLocalError
    
func1()
func2()

# 推荐做法
def func3(c):
    return c+1
    
count = func3(count)
print(count)
````

## 示例代码2.47 对比值相等（==）与对象标识相等（is）

````python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)  # True (内容一样)
print(a is b)  # False (内存地址不同)
# 因小整数缓存机制导致的例外情形
x = 101
y = 101
print(x is y)  # True
````
