## 示例代码3.3 遍历指定目录统计文件信息

```python
import os
import os.path
import time

target_dir = '/Users/xirong/workspace' # 指定待查找的目录/文件夹
res = [] # 创建一个空列表，用于收集 (修改时间戳, Word文件路径) 二元组
for root, dirs, files in os.walk(target_dir):
    for fname in files:
        if fname.endswith('.docx') is False: # 跳过非Word文件
            continue
        full_path = os.path.join(root, fname) # 获得当前文件的完整路径
        mtime = os.path.getmtime(full_path)
        res.append((mtime, full_path)) # 将修改时间放在前面，以便排序
# 遍历结束，进行排序
res.sort(reverse=True) # 激活reverse参数，让列表降序排序
# 输出结果
for (mtime, full_path) in res:
    readable_time = time.ctime(mtime) # 将时间戳转为简单易读格式
    print(full_path, readable_time)
```

## 示例代码3.4 文本文件读写

```python
input_file = "pybook-data/ch3/rawdata.txt"
output_file = "students.txt"
# 读文件
fr = open(input_file) # 以读模式（r）获得一个文件对象 
lines = fr.readlines() # 逐行读取数据
fr.close() # 关闭该文件对象
# 数据解析
lines = [x for x in lines if not x.startswith('#')] # 去掉以#开头的注释行
records = []
for line in lines:
    line = line.strip() # 去掉换行符
    elems = line.split(',') # 以逗号为分隔符，获得各个字段
    newline = ','.join(elems) # 将各个字段再以逗号为分隔符拼接回一个字符串
    records.append(newline)
# 写文件
fw = open(output_file, "w") # 以写模式（w）打开一个新的文件对象
for line in records:
    fw.write(line + '\n') # 加上换行符，否则所有记录会写到一行
fw.close()
```

## 示例代码3.5 使用csv模块将TXT转为CSV

```python
import csv
# 定义表头
headers = ["姓名", "性别", "出生日期", "学号", "所属学院", "入学年份", "籍贯", "GPA", "总学分"]
# 从students.txt中读取数据
with open('students.txt', 'r') as fr:
    lines = fr.readlines() # 读取所有行
    records = [x.strip() for x in lines if x.strip()] # 去掉换行符、空行
# 将数据写入CSV文件
with open('students.csv', 'w', newline='') as fw:
    writer = csv.writer(fw)
    # 写入表头
    writer.writerow(headers)
    # 写入数据行
    for student in records:
        writer.writerow(student.split(','))
```

## 示例代码3.6 将.csv文件中数据转存入.json文件

```python
import csv
import json
# 从students.csv中读取数据
with open('students.csv', 'r') as f:
    csv_data = list(csv.reader(f)) # 读取csv数据
    headers = csv_data.pop(0) # 从数据中将表头列表弹出
# 将每条学生记录转换为一个字典对象
students = [dict(zip(headers, row)) for row in csv_data]
# 将数据写入JSON文件
with open('students.json', 'w') as file:
    json.dump(students, file, ensure_ascii=False)
```

## 示例代码3.7 使用json模块读取JSON文件并将数据转存为HTML文件

```python
import json
with open('students.json') as fr:
    data = json.load(fr)
# 构造表头，列表为相应记录的相应字段
table = '' 
column_names = ['照片'] + list(data[0].keys())
for col in column_names:
    table += f'<th>{col}</th>'
table = f'<tr>{table}</tr>'  # 表格头是表格的第1行，以<tr>开头，</tr>结尾
# 一条学生记录对应表格的一行
table_content = []
for record in data:
    img_path = f'pybook-data/ch3/portraits/{record["学号"]}.png'
    new_row = f'<td><img src="{img_path}" width="100" height="100"></img></td>'
    new_row += ''.join([f'<td>{record[col]}</td>' for col in column_names[1:]])
    table_content.append(f'<tr>{new_row}</tr>')
table = '<table>' + table + '\n' + '\n'.join(table_content) + '</table>'
style = '<style>table, th, td {border: 1px solid black; border-collapse: collapse;}</style>'
html_content = f'<html>\n<head>{style}</head>\n<body>{table}</body>\n</html>'
# 保存为 .html 文件
with open('students.html', 'w') as fw:
    fw.write(html_content)
```

## 示例代码3.12 将HTML文档中的学生记录转存为Excel表格

```python
from bs4 import BeautifulSoup
from openpyxl import Workbook
from openpyxl.drawing.image import Image  as OpenpyxlImage # 用于在Excel文件中插入图像
from openpyxl.utils import get_column_letter #用于将数字列索引（从1开始）转换为Excel的字母列标

# 第1部分：从HTML文件中提取学生记录
html_content = open('students.html').read() # 内容如图3.2所示
soup = BeautifulSoup(html_content, 'html.parser')
table_rows = soup.find('table').find_all('tr')  # 获得HTML表格的所有行
header = table_rows[0].find_all('th') # 获得表头的所有列
column_names = [x.text for x in header] # 获取列名，即图3.2所示的"照片", "姓名", "性别"等

# 逐行遍历，将HTML表格中存储的学生记录转存到一个列表中
students_data = [] # 学生记录列表
for row in table_rows[1:]:  # 跳过第1行表头，从第2行开始解析
    cells = row.find_all('td')  # 获取当前行的所有列
    img_path = cells[0].find('img')['src'] # 获取照片文件路径
    text_data = [x.text for x in cells[1:]]   # 提取其余各字段
    record = [img_path] + text_data # 将当前学生的各项数据合并称一个列表
    students_data.append(record)  # 把当前学生记录添加到总表中

# 第2部分：将学生数据存到Excel表格中
wb = Workbook()
ws = wb.active
ws.title = "学生信息"    # 修改工作表标题
# 将数据写入Excel工作表
ws.append(column_names)   # 写入表头

# 写入学生数据并添加头像。注意从Excel表格第2行开始存数据
for row_idx, record in enumerate(students_data, start=2):  
    ws.append(record)    # 写入学生数据
    img = OpenpyxlImage(record[0])    # 在指定路径中读取图像
    img.width, img.height = 100, 100  # 设置图像尺寸
    ws[f'A{row_idx}'] = '' # 将对应单元格的字符串内容清空
    ws.add_image(img, f'A{row_idx}')  # 将图片嵌入到该单元格
    ws.row_dimensions[row_idx].height = 75 # 调整行高以容纳图片

# 创建一个字典，用于根据字段来调整列宽
column_widths = { '照片': 15, '出生日期': 12, '学号': 12, '所属学院': 12, '籍贯': 15 }
for col_idx, header in enumerate(column_names, 1): # 设置列宽
    if header in column_widths:
        col_letter = get_column_letter(col_idx) # 获取对应列索引
        ws.column_dimensions[col_letter].width = column_widths[header]

wb.save("students.xlsx") # 保存Excel文件
```

## 示例代码3.13 使用 python-docx 创建一个Word文档

```python
from docx import Document
from docx.shared import Pt # 用于设置字体大小
from docx.enum.text import WD_PARAGRAPH_ALIGNMENT # 用于设置对齐方式
from docx.oxml.ns import qn  # 用于设置中文字体

doc = Document() # 创建Word文档
# 设置全局字体为宋体（需在添加内容前定义）
doc.styles['Normal'].font.name = '宋体'
doc.styles['Normal']._element.rPr.rFonts.set(qn('w:eastAsia'), '宋体')

# 添加一级标题“注意事项”（居中对齐，加粗，字号14）
title = doc.add_heading('注意事项', level=1)
title.alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
title.runs[0].font.bold = True
title.runs[0].font.size = Pt(14)

# 添加项目符号列表
items = [ "事项1: 检查设备连接是否正常",
    "事项2: 确保数据备份已完成",
    "事项3: 操作前阅读安全手册"]
for item in items:
    para = doc.add_paragraph(style='List Bullet')  # 使用内置项目符号样式
    run = para.add_run(item)
    run.font.size = Pt(12)  # 设置列表项字号

doc.save("注意事项.docx") # 保存文档
```

## 示例代码3.14 Excel表格转Word文档

```python
from docx import Document
from docx.shared import Inches
from docx.enum.text import WD_ALIGN_PARAGRAPH
from openpyxl import load_workbook
from io import BytesIO

def load_data_from_excel(file_path):
    wb = load_workbook(file_path)
    ws = wb.active  
    # 获取表头（Excel表格第一行）
    column_names = [cell.value for cell in next(ws.iter_rows(min_row=1, max_row=1))]
    # 获取学生数据（从第二行开始），values_only=True，直接返回单元格的值
    students = [row[1:] for row in ws.iter_rows(min_row=2, values_only=True)] 
    assert(len(students) == len(ws._images)) # 照片数与学生数应相等
    images = [None]*len(students)
    for i,img in enumerate(ws._images): # 获取学生的照片
        images[i] = img._data()       
    return {'students': students, 'columns': column_names, 'images': images} # 以字典形式返回多维度数据

def save_data_to_word(records, output_path):
    column_names, students, images = records['columns'], records['students'], records['images']
    document = Document()
    document.add_heading('学生信息表', level=1) # 加入一个一级标题    
    table = document.add_table(rows=len(students) + 1, cols=len(column_names)) # 创建一个Word表格
    table.style = 'Table Grid' # 设置表格边框样式  
    for col_idx, col_name in enumerate(column_names): # 填充表头，文字居中对齐，粗体显示
        cell = table.cell(0, col_idx)
        cell.text = col_name
        para = cell.paragraphs[0]
        para.alignment = WD_ALIGN_PARAGRAPH.CENTER
        para.runs[0].font.bold = True
    for i, student in enumerate(students):  # 填充学生数据
        row_idx = i + 1       
        img_cell = table.cell(row_idx, 0)
        img_cell.text = "" # 先清空单元格，再插入图片
        para = img_cell.paragraphs[0]
        para.alignment = WD_ALIGN_PARAGRAPH.CENTER
        # 用BytesIO将二进制数据包装成“虚拟文件”，直接传递给 add_picture()
        para.add_run().add_picture(BytesIO(images[i]), width=Inches(0.8))    
        # 填充文本数据，居中对齐
        for col_idx, value in enumerate(student, start=1):
            cell = table.cell(row_idx, col_idx)
            cell.text = str(value) if value is not None else ""
            cell.paragraphs[0].alignment = WD_ALIGN_PARAGRAPH.CENTER
    # 设置列宽
    for col_idx, column in enumerate(table.columns):
        width = Inches(1.2) if col_idx == 0 else Inches(1.0)
        for cell in column.cells:
            cell.width = width
    document.save(output_path) # 保存Word文档
    
# 调用两个自定义函数，实现数据从Excel流转到Word
records = load_data_from_excel("students.xlsx")
save_data_to_word(records, "students.docx")
```

## 示例代码3.16 自定义函数来创建一个数据库表格

```python
import sqlite3 as sqlite

def create_student_table(db_path, table_name):
    conn = sqlite.connect(db_path) # 尝试连接数据库，若数据库文件不存在，则会新建一个数据库
    cursor = conn.cursor() # 游标对象
    cursor.execute(f'''CREATE TABLE IF NOT EXISTS {table_name} (
        name TEXT NOT NULL, gender TEXT NOT NULL, birth_date TEXT NOT NULL, 
        student_id TEXT PRIMARY KEY, school TEXT NOT NULL, 
        enrollment_year INTEGER NOT NULL, hometown TEXT NOT NULL, 
        gpa REAL, total_credits INTEGER)''')  # 创建表
    conn.commit() # 提交事务，让数据库执行创建表格的指令
    conn.close()  # 关闭与数据库的连接
```

## 示例代码3.17 自定义函数用于从Word文档中读取学生记录

```python
from docx import Document

def load_student_data_from_word(word_path):
    doc = Document(word_path)
    data = []
    table = doc.tables[0]
    for row in table.rows[1:]: # 跳过表头
        record = [cell.text for cell in row.cells[1:]] # 跳过第1列（照片）
        data.append(tuple(record))
    return data 
```

## 示例代码3.18 从Word文档中读取数据并存入SQLite数据库

```python
word_path = "students.docx"
db_path = "students.sqlite"
table_name = "students"
# 调用自定义函数
create_student_table(db_path, table_name)
data = load_student_data_from_word(word_path)
# 连接数据库
conn = sqlite.connect(db_path)
cursor = conn.cursor()
# 尝试往数据库中逐条插入记录
try:
    for record in data:
        cursor.execute(f'''INSERT INTO {table_name} VALUES (?,?,?,?,?,?,?,?,?)''', record) # 插入数据
    conn.commit() 
except Exception as e:
    print("发生错误，回滚事务:", e)
    conn.rollback()  # 仅在没有 commit 时有效
finally:
    conn.close()
```
