## 示例代码2.10 使用for循环遍历一个列表

```python
data = [("李四", 92), ("张三", 86), ("王五", 82)]
for x in data:
    name = x[0]
    score = x[1]
    print(name, score) # 逐行输出名字和分数

for (name, score) in data:
    print(name, score) # 逐行输出名字和分数

for rank, (name,score) in enumerate(data, start=1):
    print(name, score, rank) # 逐行输出名字、分数和排名

# 通过遍历找出最高分
best_score = -100 # 用于记录遍历过程中找到的最高分
best = -1 # 用于记录最高分学生在列表中的位置
for i, x in enumerate(data):
    if x[1] > best_score: # 找到更高分，更新
        best_score = x[1]
        best = i
print("最高分：", data[best]) # 最高分：("李四", 92)
```

## 示例代码2.14 使用while循环找出列表中满足特定条件的元素

```python
data = [("李四", 92), ("王五", 82), ("张三", 86)]

i = 0 # 位置索引，从0开始
while (i < len(data)) and (data[i][1] > 85): # 组合式条件
    print(data[i]) # 输出学生信息
    i = i + 1 # 更新索引，用于访问列表中的下一个元素
```

## 示例代码2.15 使用break与continue改变循环流程
注意观察break与continue的缩进

```python
data = [("李四", 92), ("张三", 86), ("王五", 82)]

# 仅输出列表中第一个分数大于85的元素
for (name,score) in data:
    if score > 85:
        print(name, score)
        break # 循环结束
print('----------------')      
# 输出列表中所有分数大于85的元素
for (name,score) in data:
    if score <= 85:
        continue # 跳过循环体的剩余部分代码
    print(name, score)
```

## 示例代码2.17 带有异常处理机制的除法运算

```python
try:
    num1 = int(input("请输入一个整数作为被除数: "))
    num2 = int(input("请输入一个整数作为除数: "))
    result = num1 / num2
    print(f"{num1} / {num2} = {result}")
except ValueError:
    print("【错误提示】请输入有效的整数")
except ZeroDivisionError:
    print("【错误提示】除数不能为零")
except Exception as e:
    print(f"【未知错误】{e}")
finally:
    print("*** 程序结束 ***")
```

## 示例代码2.19 读取列表中的元素

```python
# 使用索引访问单个元素
my_list = ['a', 'b', 'c', 'd', 'e']
print(my_list[0]) # 'a' - 第一个元素
print(my_list[1]) # 'b' - 第二个元素
print(my_list[2]) # 'c' - 第三个元素
print(my_list[-1]) # 'e' - 倒数第一个元素
print(my_list[-2]) # 'd' - 倒数第二个元素
print(my_list[-3]) # 'c' - 倒数第三个元素
# 使用切片访问子序列
print(my_list[1:3]) # ['b', 'c'] - 索引1到2(不包括3)
print(my_list[:3]) # ['a', 'b', 'c'] - 从头至索引2
print(my_list[3:]) # ['c', 'd'] - 从索引3到末尾
print(my_list[::2]) # ['a', 'c', 'e'] - 间隔取元素
print(my_list[::-1]) # ['e', 'd', 'c', 'b', 'a']- 反转列表
```

## 示例代码2.20 集合常用操作

```python
# 创建集合
s = set() # 空集合
s = set([1, 2, 3, 4]) # 从列表创建 {1, 2, 3, 4}
s = set([1, 2, 3, 3]) # {1, 2, 3} 注意元素3已被去重
s = {1, 2, 3, 4} # 直接使用花括号

# 添加元素
s.add(5) # 添加一个元素， 集合变为{1, 2, 3, 4, 5}
s.update([6, 7]) #添加多个元素，集合变为{1, 2, 3, 4, 5, 6, 7}

# 删除元素
s.remove(7) # 如果元素不存在会报告KeyError
s.discard(10) # 安全移除，不存在也不会报错

# 成员测试
print(1 in s) # 输出 True
print(7 in s) # 输出 False

# 获取集合大小，即集合中元素的个数
len(s) # 6

# 清空集合
s.clear()

# 将集合变为列表
s = {1, 2, 3, 4, 3} # 集合会自动去重，产生集合 {1, 2, 3, 4}
t = list(s) # 产生列表 [1, 2, 3, 4]
```

## 示例代码2.21 字典常用操作

```python
# 创建字典
d = {} # 空字典
d = dict() # 空字典
d1 = {'name': 'Alice', 'age': 25} # 直接初始化
d2 = dict(name='Alice', age=25) # 通过关键字参数初始化
d3 = dict([('name','Alice'), ('age', 25)]) # 从键值对序列创建
d4 = dict.fromkeys(['a', 'b', 'c'], 0) # 从键列表创建默认值字典 {'a': 0, 'b': 0, 'c': 0}

# 以键（key）为索引来访问元素
print(d1['name']) # 'Alice' (键不存在会报KeyError)
print(d1.get('age')) # 25 (安全访问，不存在返回None)
print(d1.get('gender', 'unknown')) # 不存在则返回指定值'unknown'

# 检查键（key）是否存在
print('name' in d2) # 输出 True
print('Alice' in d2) # 输出 False

# 添加/修改元素
d1['gender'] = 'female' # 添加新键值对
d1['age'] = 26 # 修改已有键的值

# 删除元素
del d1['age'] # 删除键值对
value = d1.pop('name') # 删除并返回值
d1.clear() # 清空字典

# 通过items())遍历字典
for key, value in d4.items():
    print(f"{key}:{value}") # 逐行输出 a:0, b:0, c:0

# 通过keys()遍历字典
for key in d4.keys():
    print(f"{key}:{d4[key]}") # 逐行输出 a:0, b:0, c
```

## 示例代码2.22 利用特殊字典类型Counter统计字母在序列中的出现频率

```python
from collections import Counter
a = ['a', 'b', 'b', 'c', 'c', 'c']
c = Counter()
for i,x in enumerate(a):
    c[x] = c[x] + 1
    print(f'iter-{i}', c.most_common()) 
```
