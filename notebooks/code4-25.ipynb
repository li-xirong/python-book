# Pandas 表格拼接与合并

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'姓名': ['张三', '李四'], '语文': [85, 92]})
df2 = pd.DataFrame({'姓名': ['王五', '赵六'], '语文': [88, 95]})
df3 = pd.DataFrame({'姓名': ['李四', '张三'], '数学': [90, 75]})
df4 = pd.concat([df1, df2], axis=0, ignore_index=True) # 纵向拼接
df5 = pd.concat([df1, df3['数学']], axis=1) # 横向拼接
df6 = pd.merge(df1, df3, on='姓名')
df1, df2, df3, df4, df5, df6
```
