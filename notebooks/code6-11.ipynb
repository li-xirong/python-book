# 绘制各专业文理科招生人数对比的堆叠柱状图

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS', 'Heiti TC', 'Hiragino Sans GB', 'PingFang SC'] # 正常显示中文
plt.rcParams['axes.unicode_minus'] = False # 正常显示负号
```

```python
majors = ['经济学','新闻传播学','人力资源管理','国际政治','外国语言文学'] 
wenke = [7, 5, 6, 4, 5] # 文科招生人数
like = [9, 2, 5, 3, 5] # 理科招生人数
fig, axes = plt.subplots()
x = np.arange(len(majors))
wenke_bar = axes.bar(x, wenke, color='skyblue', width=0.6, label='文科') 
like_bar = axes.bar(x, like, color='orange', width=0.6, bottom=wenke, label='理科')
axes.bar_label(wenke_bar, label_type='center', fontsize=12) 
axes.bar_label(like_bar, label_type='center', fontsize=12) 
axes.set_xlabel("专业", fontsize=14)
axes.set_ylabel("招生人数", fontsize=14) 
axes.set_ylim(0, 17) 
axes.set_yticks(ticks=range(17)) 
axes.set_xticks(ticks=range(5)) 
axes.set_xticklabels(majors) 
axes.grid(True, axis='y') 
axes.legend(fontsize=12)
```
