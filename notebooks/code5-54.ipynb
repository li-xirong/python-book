## 创建BertDataset和DataLoader对象，用于加载训练数据

```python
from torch.utils.data import DataLoader, Dataset
from p5lib.ch5 import load_imdb_data, load_bert_model, BertDataset

sents_train, sents_test, y_train, y_test = load_imdb_data()
_, tokenizer = load_bert_model()
```

```python
train_ds = BertDataset(sents_train, y_train, tokenizer)
train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
```
