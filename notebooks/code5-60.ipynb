# 从HuggingFace镜像批量下载基座模型

```python
from huggingface_hub import snapshot_download

endpoint = 'https://hf-mirror.com'
model_list = [("sentence-transformers/all-MiniLM-L6-v2", "fm/all-MiniLM-L6-v2"), ("moka-ai/m3e-small", "fm/m3e-small"), ("sentence-transformers/clip-ViT-B-32", "fm/clip-ViT-B-32")]
for repo_id, local_dir in model_list:
    snapshot_download(repo_id=repo_id, local_dir=local_dir, endpoint='https://hf-mirror.com')
```
