# DataFrame中的数据添加与删除

```python
import pandas as pd
import numpy as np
```

```python
dates = pd.date_range("20200901", periods=3)
data = [[12, 20, "晴"], [9, 18, "小雨"], [12, 23, "多云"]]
df = pd.DataFrame(data, index=dates, columns=["low-temp", "high-temp", "weather"])
```

```python
# 添加两条（行）新记录
df.loc["2020-09-04"] = [10, 22, "小雨"]
df.loc["2020-09-05"] = [10, 24, np.nan] # np.nan表示缺失值
print("添加两条（行）新记录\n", df)

# 连续添加记录
for i in range(5):
    df.loc[len(df)] = [11, 25, "阴"]
print("连续添加记录\n", df)

# 添加一列
df['温差'] = df['high-temp'] - df['low-temp']
print("添加一列\n", df)

# 原地删除存在缺失值的记录（对应2020-09-05索引的行）
df.dropna(inplace=True) 
print("原地删除存在缺失值的记录\n", df)

# 原地删除新增加的列
df.drop(columns=['温差'], inplace=True)
print("原地删除新增加的列\n", df)

# 原地删除重复记录
df.drop_duplicates(inplace=True)
print("原地删除重复记录\n", df)

# 原地删除最后一行
df.drop(df.index[-1], inplace=True)
print("原地删除最后一行\n", df)
```
