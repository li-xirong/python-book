### 示例代码 6.14 绘制散点图 

````python
points = np.random.rand(50, 2) # 随机生成50个二维样本点
fig, axes = plt.subplots()
axes.plot(points[:,0], points[:,1], 
          linestyle='none', # 隐藏数据点之间的连线
          marker='o', # 标记形状（圆形）
          markersize=20, # 标记大小
          markerfacecolor='red', # 标记填充色
          markeredgecolor='blue', # 边缘颜色
          markeredgewidth=2, # 边缘宽度
          alpha=0.5 # 填充色透明度
         )
````

### 示例代码 6.15 绘制带分界线的散点图 

````python
above = points[points[:,0]<points[:,1]] # 满足 x<y 的点落在直线 y=x 上方
below = points[points[:,0]>points[:,1]] # 满足 x>y 的点落在直线 y=x 下方

fig, axes = plt.subplots()

axes.plot(above[:,0], above[:,1], linestyle='none', marker='o', markersize=20, markerfacecolor='red', 
          markeredgecolor='blue', markeredgewidth=2, alpha=0.5, label="在$y=x$上方") # 上方的点用红色填充

axes.plot(below[:,0], below[:,1], linestyle='none', marker='o', markersize=20, markerfacecolor='yellow', 
          markeredgecolor='blue', markeredgewidth=2, alpha=0.5, label="在$y=x$下方") # 下方的点用黄色填充

x = np.linspace(0,1,50)
y = x
axes.plot(x, y, '--', color="gray") # 绘制直线 y=x

plt.legend(fontsize=12)
````

### 课堂练习6-1-3

给定由np.random.rand(50,2)随机生成的50个二维数据点，绘制散点图:
+ 每个点的标记样式仍沿用示例代码，但要求距离直线$y=x$越远的点，透明度越高（即存在感越低）;
+ 对于一个数据点$(u,v)$，其到$y=x$的距离记为$d$，则有$d=\frac{|u-v|}{\sqrt{2}}$;
+ 该点对应的透明度参数$\mbox{alpha}=(1-\frac{d}{d_{max}})^2$，其中$d_{max}$为距离最大值，即$\frac{1}{\sqrt{2}}$。

### 双坐标 (twin-axes)

````python
fig, ax_left = plt.subplots()

x = np.linspace(0, 2, 101)
ax_left.plot(x, x**2, lw=2, c="green")
ax_left.set_ylabel(r"area $(m^2)$", fontsize=18, color="blue")
ax_left.tick_params(axis='y', colors='green')
    
ax_right = ax_left.twinx()
ax_right.plot(x, x**3, lw=2, color="red")
ax_right.set_ylabel(r"volume $(m^3)$", fontsize=18, color="red")
ax_right.tick_params(axis='y', colors='red')

ax_left.set_xlabel('shared x', fontsize=18)

plt.show()
````

### 定制坐标
通过调整（隐藏、移动）坐标轴的边界线 (spines)，让原点成为图的中心点

````python
fig, ax = plt.subplots()

ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

ax.spines['bottom'].set_position(('data',0)) # set position of x spine to x=0
ax.spines['left'].set_position(('data',0))   # set position of y spine to y=0

ax.xaxis.set_ticks_position('bottom')

t = np.linspace(-1, 1., 200)
y1 = [x**2 for x in t]
y2 = [x**3 for x in t]

ax.plot(t, y1, 'b', lw=2, label='$y=x^2$');
ax.plot(t, y2, 'r', ls='--', lw=2, label='$y=x^3$')
ax.legend()
````

### 多个子图

````python
fig, axes = plt.subplots(nrows=1, ncols=3)
colors = 'r b g k y'.split()

x = np.linspace(0, 1, 50)
y = x**2

for i,ax in enumerate(axes):
    ax.plot(x, y, colors[i])
    ax.set_xlabel('x', fontsize=18)
    ax.set_ylabel('y')
    ax.set_title('subplot %d' % i)

fig.tight_layout() #自动调整子图位置使它们不重叠
````

````python
fig, axes = plt.subplots(nrows=2, ncols=3)
max_row, max_col = axes.shape

colors = 'red yellow orange purple green black blue'.split()
for i in range(max_row):
    for j in range(max_col):
        ax = axes[i][j]
        index = i * max_col + j
        ax.plot(x, y, color=colors[index])
        ax.set_xlabel('x')
        ax.set_ylabel('y')
        ax.set_title('subplot (%d,%d)' %(i,j))
    
fig.tight_layout()
````

### 制作动图
两步法
+ Step 1. 利用matplotlib画出每一帧画面
+ Step 2. 利用imageio将帧序列存为gif动图

````python
import os
import imageio.v3 as iio

img_dir = 'bigdata'
os.makedirs(img_dir, exist_ok=True)
img_paths = []
n = 11
x = np.linspace(-5, 5, n)

fig, axes = plt.subplots(figsize=(5, 5)) 
axes.axis([-6, 6, -150, 150])
plt.grid()
for i in range(1, n+1):
    x1 = x[:i]
    y1 = x[:i]**3
    axes.plot(x1, y1, marker='o', ms=10,  mew=3, mec='k', mfc='none', lw=5, color='#CD5555')
    img_fname = os.path.join(img_dir, "%d.jpg"%i)
    print(img_fname)
    plt.savefig(img_fname)
    img_paths.append(img_fname)

frames = np.stack([iio.imread(x) for x in img_paths], axis=0)
iio.imwrite('test.gif', frames, fps=2)
````
