## 示例代码4.24 DataFrame中的数据添加与删除

````python
# 添加两条（行）新记录
df.loc["2020-09-04"] = [10, 22, "小雨"]
df.loc["2020-09-05"] = [10, 24, np.nan] # np.nan表示缺失值
# 连续添加记录
for i in range(5):
    df.loc[len(df)] = [11, 25, "阴"]
# 添加一列
df['温差'] = df['high-temp'] - df['low-temp']
# 原地删除存在缺失值的记录（对应2020-09-05索引的行）
df.dropna(inplace=True) 
# 原地删除新增加的列
df.drop(columns=['温差'], inplace=True)
# 原地删除重复记录
df.drop_duplicates(inplace=True)
# 原地删除最后一行
df.drop(df.index[-1], inplace=True)
df
````

## 示例代码4.26 人民大学在北京市各专业最低录取分历年数据汇总

````python
# Step 1. 读取每年的录取分数表
years = [2020, 2021, 2022, 2023]
dfs = {}
for year in years:
    csv_file = f'pybook-data/ruc_data/北京市/{year}.csv'
    dfs[year] = pd.read_csv(csv_file)
    
# Step 2. 按专业合并最低分
columns = ['专业名称', '最低分']
merged_bj = dfs[years[0]][columns].rename(columns={'最低分': years[0]})
for year in years[1:]:
    sub_table = dfs[year][columns].rename(columns={'最低分': year})
    merged_bj = pd.merge(merged_bj, sub_table, on='专业名称')
    
# Step 3. 对各专业按照分数由新到旧、由高到低排序    
merged_bj.sort_values(years[::-1], ascending=[False]*len(years), ignore_index=True, inplace=True)

merged_bj
````

## 示例代码4.27 计算各专业相比当年最低分平均值的偏差

````python
# Step 1. 按列取平均（跳过第1列"专业名称"）
mean_s = merged_bj.iloc[:, 1:].mean() 
# Step 2. 计算偏差 
diff = merged_bj.iloc[:, 1:].subtract(mean_s)
# Step 3. 观察结果
out = pd.concat([merged_bj[['专业名称']], diff.add_suffix('_偏差')], axis=1)
with pd.option_context('display.float_format', '{:.1f}'.format): # 临时设置浮点格式
    print(out)
````

## 示例代码4.29 人民大学各专业填报热度分析

````python
from pathlib import Path

# Step 1. 获得所有省份2023年的CSV文件列表
root_path = Path('pybook-data/ruc_data')
csv_files = list(root_path.rglob('2023.csv'))
csv_files = [str(x) for x in csv_files]

# Step 2. 读入各省的招生数据表并清洗数据
tables = [None]*len(csv_files)
for i,csv_fname in enumerate(csv_files):
    prov = csv_fname.split('/')[-2]
    df = pd.read_csv(csv_fname).rename(columns={'理工/物理类 最低':'最低分'})  
    sub_table = df[['专业名称','最低分']].drop_duplicates() # 只统计理工类，去掉可能存在的重复记录 
    sub_table.insert(0, '省份', [prov]*len(sub_table))
    tables[i] = sub_table.dropna() # 去掉没有理工类的记录
    
# Step 3. 将各省各专业投档分纵向拼接成一个长表
long_table = pd.concat(tables, ignore_index=True)
print('记录数:', len(long_table)) # 记录数: 537

# Step 4. 透视重组，获得一个宽表，每行代表一个省份，每列对应一个专业
wide_table = long_table.pivot(index='省份', columns='专业名称', values='最低分')

# Step 5. 专业热度统计
#wide_table.fillna(0, inplace=True) #先用0填充缺失值
wide_table['最高分专业'] = wide_table.idxmax(axis=1) #找出每个省（直辖市）投档线最高的专业
wide_table['最高分'] = wide_table.iloc[:, :-1].max(axis=1) #单独列出最热门专业的投档分

wide_table[['最高分专业', '最高分']].head(10)
````

````python
wide_table['最高分专业'].value_counts().to_frame() # 统计热门专业的省份频率
````

## 示例代码4.30 按地区频度校正之后的专业热度估计

````python
s1 = wide_table['最高分专业'].value_counts()
s2 = (wide_table.iloc[:,:-2]>0).sum()
s1.name = '热度'
s2.name = '地区频度'

overview = pd.concat([s1, s2], axis=1).dropna()
overview['占比'] = overview[s1.name] / overview[s2.name]
overview.sort_values(by='占比', ascending=False)
````

### 示例代码4.32 抓取2024年北京高考数据

````python
import os
from p3lib.ch3 import url_to_filename, PageDownloader

data_dir = 'beijing'
crawler = PageDownloader(save_dir=data_dir)
#抓取北京市2024年本科普通批录取投档线
target_url = "https://gaokao.eol.cn/bei_jing/dongtai/202407/t20240720_2625168.shtml"
crawler.process(target_url)
score_file = os.path.join(data_dir, f'{url_to_filename(target_url)}.html')
assert(os.path.exists(score_file)) #检查目标网页是否已保存到指定路径
#抓取2024年北京高考分段表
target_url = "https://gaokao.eol.cn/bei_jing/dongtai/202406/t20240625_2619192.shtml"
crawler.process(target_url)
rank_file = os.path.join(data_dir, f'{url_to_filename(target_url)}.html')
assert(os.path.exists(rank_file))
````

### 示例代码4.33 投档线数据清洗

````python
score_df = pd.read_html(score_file)[0]
score_df = score_df[1:].reset_index(drop=True) #跳过原表第一行，并重置行索引
score_df.columns = ['院校代码', '院校', '专业组代码', '限制', '投档线'] #列重命名
score_df['投档线'] = score_df['投档线'].astype(int) 
score_df
````

### 示例代码4.34 统计本科普通批次各高校专业组数量

````python
univ_df = score_df['院校'].value_counts().to_frame().reset_index()
univ_df.columns = ['院校', '专业组数量']
print(f'普通批次共有{len(univ_df)}家高校、{len(score_df)}个专业组') # 普通批次共有591家高校、1258个专业组
major_group = score_df['限制'].value_counts().to_frame().reset_index() #统计专业填报限制情况
major_group
````

### 示例代码4.35 分段表清洗并规整

````python
rank_df = pd.read_html(rank_file)[0]
rank_df = rank_df[1:].reset_index(drop=True)
rank_df.columns = ['分数', '人数', '位次']
rank_df.iloc[0,0]= "700" #将第一行的700-750替换为700
hit = rank_df['分数'].str.contains(r'本科线', na=False) #正则匹配，查找含有'本科线'字样的行
cutoff = hit.idxmax() #定位本科线对应的行
rank_df = rank_df[:cutoff+1] # 舍弃该行之后的数据
rank_df['分数'] = rank_df['分数'].str.replace(r'（.*?）', '', regex=True) #清除注释（特招线、本科线）
rank_df = rank_df.astype(int)
rank_df
````

### 示例代码4.36 分数转位次

````python
score_rank = rank_df.set_index('分数')['位次'] # 先构造“分数->位次”的映射表
score_df['位次'] = score_df['投档线'].map(score_rank)
score_df
````

### 示例代码4.37 查找指定位次区间的专业组

````python
n_24, n_25 = 6.72, 7.89 # 2024, 2025年北京高考人数（单位：万）
rank_25 = 16470 # 考生当前位次
query = int(rank_25*n_24/n_25+0.5) # 按人数比值修正位次
error = 0.05                      
low  = int(query * (1 - error))
high = int(query * (1 + error))
hit = score_df[(score_df['位次'] >= low) & (score_df['位次'] <= high)]
print(f'2025年位次: {rank_25}，对应2024年位次: {query}')
print(f'位次在[{low}, {high}]的专业组数：{len(hit)}')
"""输出：
2025年位次: 16470，对应2024年位次: 14028
位次在[13326, 14729]的专业组数：47
"""
sub = hit[hit['限制'].str.contains(r'不限') == True]
print(f'无限制的专业组数：{len(sub)}')
sub = hit[hit['限制'].str.contains(r'物理$') == True] 
print(f'限"物理"的专业组数：{len(sub)}')
````
