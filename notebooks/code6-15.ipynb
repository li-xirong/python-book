# 绘制二维随机散点图（定制）

散点分组显示、添加图例

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS', 'Heiti TC', 'Hiragino Sans GB', 'PingFang SC'] # 正常显示中文
plt.rcParams['axes.unicode_minus'] = False # 正常显示负号
```

```python
# 若想每次结果都固定: 
# seed = 24
# np.random.seed(seed)

points = np.random.rand(50, 2)
above = points[points[:,0]<points[:,1]]
below = points[points[:,0]>points[:,1]]
fig, axes = plt.subplots()
axes.plot(above[:,0], above[:,1], linestyle='none', marker='o', 
          markersize=20, markerfacecolor='red', markeredgecolor='blue', 
          markeredgewidth=2, alpha=0.5, label="在$y=x$上方") 
axes.plot(below[:,0], below[:,1], linestyle='none', marker='o', 
          markersize=20, markerfacecolor='yellow', markeredgecolor='blue', 
          markeredgewidth=2, alpha=0.5, label="在$y=x$下方") 
x = np.linspace(0,1,50)
y=x
axes.plot(x, y, '--', color="gray") 
plt.legend(fontsize=12)
```
