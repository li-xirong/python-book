## 文本数据清洗

````python
data = [" 2026-08-25_ID98453_John Doe _MEAL_ $128.50USD ",
"2026/08/26-ID10247-Marta- hotel -$450.00USD",
"2026.08.26_id00382_ ALEX _flight_$1,200.00 USD"]
data
````

````python
records = []
for line in data:
    refined = line.strip().replace('-','_').replace('/','_').replace('.','_')
    elems = refined.split('_')
    elems = [x.strip() for x in elems if x.strip()]
    print(elems)
    date = '-'.join(elems[:3])
    trans_id = elems[3].lower().replace('id','')
    name = elems[4]
    category = elems[5]
    cost = '.'.join(elems[-2:]).replace('USD','').replace('$','').replace(',','')
    cost = float(cost)
    records.append((trans_id, date, name, category, cost))  
print('')
for rec in records:
    print(rec)
````

## 示例代码2.25 字符串分割（split）、拼接（join）与连接（concat）

````python
text = "中国人民大学,北京|重庆大学,重庆|浙江大学,杭州"
# 先按 "|" 分割每条记录
records = text.split("|")
# 遍历每条记录，再按 "," 分割字段
for record in records:
    univ, city = record.split(",")
    out = "".join([univ, "位于", city])
    out1 = f"{univ}位于{city}" # out1是out的f-string实现
    print(out)
    print("-"*20)
````

## 示例代码2.30 带参数、有返回值的函数

````python
def greet1(name):
    return f"Hello, {name}"

def greet2(name: str) -> str:
    return f"Hello, {name}"

def greet3(name, title='Dr.'):
    return f"Hello, {title} {name}"

def greet4(name: str, title: str='Dr.') -> str:
    return f"Hello, {title} {name}"
# 调用函数
print(greet1("Watson")) # 输出: Hello, Watson
print(greet2("Watson")) # 输出: Hello, Watson
print(greet3("Watson")) # 输出: Hello, Dr. Watson
print(greet4("Watson")) # 输出: Hello, Dr. Watson
print(greet4("Watson", 'Mr.')) # 输出: Hello, Mr. Watson
print(greet4('Mr.', "Watson")) # 输出: Hello, Watson Mr.
print(greet4(title="Mr.", name="Watson")) # 输出: Hello, Mr. Watson
````

## 示例代码2.31 将百分制分数转换为4 分制绩点

````python
def calc_linear_gpa(score):
    if score < 60:
        return 0.0
    elif score > 90:
        return 4.0
    gpa = (score - 50) / 10
    return round(gpa, 2)
# 测试代码
for score in range(50, 100, 5):
    print(f'{score} -> {calc_linear_gpa(score)}')
````

## 示例代码2.32 形参与实参同名（但相互独立）

## 示例代码2.33 利用lambda操作实现列表的多键排序

````python
data = [
    {'name': 'Alice', 'age': 25, 'score': 90},
    {'name': 'Bob', 'age': 30, 'score': 95},
    {'name': 'Charlie', 'age': 25, 'score': 95},
    {'name': 'David', 'age': 30, 'score': 80}
]
# 先按score降序，score相同情况下再按age升序
sorted_data = sorted(data, key=lambda x: (-x['score'], x['age']))
for item in sorted_data:
    print(item)
````
