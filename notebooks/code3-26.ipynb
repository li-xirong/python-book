# 使用 DrissionPage 登录民宿网站并读取数据

```python
from DrissionPage import ChromiumOptions, ChromiumPage

# 第一步，启动浏览器
co = ChromiumOptions()
bro = ChromiumPage(co)
bro.set.auto_handle_alert() # 自动确认弹窗
# 第二步，向目标 URL 请求网页
bro.get("https://m.tujia.com/hotel_city73/7-2,101/")
input("按回车键继续...") # 完成登录操作以后，按下回车键，程序继续运行
# 第三步，解析网页并提取所需信息
item = bro.ele(".unit-item-padding")
item_link = item.ele("tag:a")
print(item_link.attr("title")) # 获取标题
print(item_link.attr("href")) # 获取详情页链接
```
