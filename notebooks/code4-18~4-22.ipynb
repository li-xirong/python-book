# 4.18 创建Series对象的不同方式

```python
import pandas as pd
import numpy as np
```

```python
#使用列表的方式创建Series，内容为2023年人大理科各专业在福建的最高录取分数
scores_list = [671, 685, 676, 686, 672, 672, 672, 678, 679, 674, 680, 672, 671, 653, 642]
s0 = pd.Series(scores_list)

#使用Numpy的方式创建Series，内容同上
scores_numpy = np.array([671, 685, 676, 686, 672, 672, 672, 678, 679, 674, 680, 672, 671, 653, 642])
s1 = pd.Series(scores_numpy)

#使用字典的方式创建Series
scores_dict = {"外国语言文学类":671, "金融学类":685,"经济学类":676,"法学":686,"工商管理类":672,
                "社会科学试验班（大数据技术与农林经管双主学位项目）":672,
                "社会科学试验班（应用经济与农村发展管理双主学位项目）":672,"统计学类":678,
                "人工智能（拔尖班）":679,"计算机类":674,"计算机类（图灵实验班）":680,
                "数学类（英才班）":672,"应用心理学":671,"金融学（中外合作办学）":653,
                "人力资源管理（中外合作办学）":642}
s2 = pd.Series(scores_dict)

#使用索引的方式创建Series
scores_data = [671, 685, 676, 686, 672, 672, 672, 678, 679, 674, 680, 672, 671, 653, 642]
scores_index = ["外国语言文学类", "金融学类","经济学类","法学","工商管理类",
                "社会科学试验班（大数据技术与农林经管双主学位项目）",
                "社会科学试验班（应用经济与农村发展管理双主学位项目）","统计学类",
                "人工智能（拔尖班）","计算机类","计算机类（图灵实验班）",
                "数学类（英才班）","应用心理学","金融学（中外合作办学）",
                "人力资源管理（中外合作办学）"]
s3 = pd.Series(scores_data, index=scores_index)
print(s0, s1, s2, s3)
```

# 4.19 读取Serie对象中的数据 

```python
print(s2.values)    #查看所有值
print(s2.index)     #查看所有索引
print(s2[10])       #位置索引，查看法学专业分数
print(s2[[3, 10]])  #位置索引列表，查看法学和计算机类
print(s2["法学"])    #标签索引，查看指定专业的分数线
print(s2[["法学", "计算机类（图灵实验班）"]]) #标签索引列表，查看多个专业的分数线
```

# 4.20 非数值型Series，其values可同时包含不同数据类型

```python
s4 = pd.Series(['E.B.', 'White', 'Male', 'American', 1899, 1985])
s5 = pd.Series(['E.B.', 'White', 'Male', 'American', 1899, 1985], 
              index=["First-Name", "Family-Name", "Gender", "Nationality", "Born", "Died"])
print(s4)
print(s5)
```

# 4.21 调用Series.describe() 进行数据统计

```python
d = s2.describe() 
print(d['count'], d['mean'], d['min'], d['max']) # 输出 15.0 672.2 642.0 686.0
t = pd.Series(['t1', 't1', 't1', 't2', 't3', 't3'])
d = t.describe()
print(d)
```

# 4.22 Series数据筛选

```python
d = s2.describe()
print(s2[s2>d['mean']])
```
