# 大模型生成的猜数字游戏代码

```python
# 猜数字游戏（人猜电脑想）
import random
target = random.randint(1, 1000)   # 电脑想的数
tries = 0                          # 计数器
while True:
    guess = int(input("请输入 1-1000 之间的整数："))
    tries += 1
    if guess == target:
        print(f"恭喜，猜对了！共猜了 {tries} 次。")
        break
    print("大了" if guess > target else "小了")
```
